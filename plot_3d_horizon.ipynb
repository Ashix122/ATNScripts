```python
import numpy as np
import plotly.graph_objs as go # type: ignore
import plotly.io as pio
line_width=3
```

```python
import numpy as np

horizon_path = "/home/ashiquetnizar/Projects/Reps/cc_axi/findAH/horizon_4.114.txt"

data = np.loadtxt(horizon_path, comments="#")

x_vals = data[:, 0]
z_vals = data[:, 1]
bifurcated = True
```

```python
fig = go.Figure()

x = x_vals
z = z_vals

# reflect about z-axis (build one closed loop in the +z half)
# drop endpoints to avoid duplicating the axis seam
x0 = x
z0 = z
x_m = -x0[1:-1]
z_m =  z0[1:-1]

x_upper = np.concatenate([x0, x_m[::-1]])
z_upper = np.concatenate([z0, z_m[::-1]])

# close upper loop
x_upper = np.append(x_upper, x_upper[0])
z_upper = np.append(z_upper, z_upper[0])

# add upper loop
fig.add_trace(go.Scatter(
  x=x_upper, y=z_upper, mode="lines",
  # name="Horizon",
  showlegend=False,
  line=dict(width=line_width)
))

if bifurcated:
  # lower loop is just x-axis reflection of the upper loop
  x_lower = x_upper
  z_lower = -z_upper

  # add lower loop as a second trace, but hide legend
  fig.add_trace(go.Scatter(
    x=x_lower, y=z_lower, mode="lines",
    showlegend=False,
    line=dict(width=line_width)
  ))

fig.update_layout(
  title="Horizon",
  template='plotly_white',     # ← white background stylexaxis_title="x",
  yaxis_title="z",
  xaxis_title="x",
  # legend_title="File",
  width=1200,
  height=1200,
  dragmode="pan",
  xaxis=dict(scaleanchor="y", scaleratio=1),
  yaxis=dict(scaleratio=1),
  font=dict(
    family="Garamond, Georgia, serif",
    size=20,
    color="black"
  ),
)


fig.show()
```

```python
phi = np.linspace(0, 2*np.pi, 100)
#pio.renderers.default = "browser"

X = np.outer(x_vals, np.cos(phi))
Y = np.outer(x_vals, np.sin(phi))
Z = np.outer(z_vals, np.ones_like(phi))

fig = go.Figure(data=[
    go.Surface(
        x=X,
        y=Y,
        z=Z,
        colorscale='Portland',
        showscale=False
    )
])

if bifurcated:
    fig.add_trace(
    go.Surface(
        x=X,
        y=Y,
        z=-Z,
        colorscale='Portland',
        showscale=False
    )
)


fig.show()
```
